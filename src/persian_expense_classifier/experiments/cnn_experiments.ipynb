```python
from persian_expense_classifier.data.data_loader import loadData
from keras.layers import TextVectorization
from persian_expense_classifier.preprocessing.input_standardize import input_standardize
```

```python
X, y = loadData()

X.shape, y.shape
```

```python
text_vectorize = TextVectorization(
    output_sequence_length=20,
    standardize=input_standardize,
)


text_vectorize.adapt(X)

text_vectorize.vocabulary_size()
```

```python
result = text_vectorize.call("غذا خریدم 500 تومن")

result
```
